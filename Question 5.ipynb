##### Question 5 -  Write a function that gets a matrix and normalizes the matrix so that every row sums to 1.

```python
import numpy as np
```

```python
def normalised_matrix(matrix): #define a normalised_matrix() function with matrix as a parameter
    sum_of_each_rows = matrix.sum(axis=1, keepdims=True) #use sum() function to calculate the sum of each row and keepdims= True ensures that the result retains the same number of dimensions as the original array because when we sum over an axis, NumPy reduces the number of dimensions by default.
    normalized = matrix/ sum_of_each_rows #normalize each row by dividing each element in a matrix by the corresponding row sum
    return normalized

matrix = np.array([
          [1,2,3,4],
          [2,4,6,8],
          [1,3,5,7]
         ])
normalized = normalised_matrix(matrix)
print("Normalized matrix:\n" ,normalized)
```
