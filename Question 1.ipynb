#### Question 1- Create a 10 × 10 array with random integers in the range 1-100 and print the following:

```python
import numpy as np
```

```python
arr = np.random.randint(1, 101, size=(10,10)) #creates a 10 x 10 array by generating random integers from 1 to 100.
print(arr)
```

#### (a) The minimum and maximum values in the array.

```python
minimum_value= arr.min() #computes minimum value in the array by using min() function
print("Minimum value:", minimum_value)
```

```python
maximum_value= arr.max() #computes maximum value in the array by using max() function.
print("Maximum value:", maximum_value)
```

#### (b) The average value of every even row (rows 0, 2, ..., 8).

```python
even_rows = arr[::2] #selecting every even rows 0,2,4,6,8
print("Every even row:\n", even_rows) #Displays every even row in the array
avg_of_even_row = np.mean(even_rows, axis=1) #using mean() function to calculate the average value of every even row
print("Average of every even row: ", avg_of_even_row)
```

#### (c) The standard deviation in every odd column (columns 1, 3, ..., 9)

```python
odd_col = arr[: , 1::2] #selecting every odd columns 1,3,5,7,9
print("Every odd column:\n" ,odd_col) #displays every odd column in the array
std_dev = np.std(odd_col, axis=0) #using std() function to calculate the standard deviation in every odd column
print("Standard deviation of every odd column:",std_dev)
```

#### (d) How many rows contain a number less than 5?

```python
rows= np.any(arr<5, axis=1) #using any() function to access rows that contain a number less than 5
row_count= np.sum(rows) #using sum() function to calculate the sum of total rows that contain a number less than 5
print("Total number of rows that contain a number less than 5: " ,row_count)
```

#### (e) Which columns contain a number greater than 90? (print the indexes of the columns)

```python
cols= np.any(arr>90, axis=0) #using any() function to access columns that contain a number greater than 90
indices= np.where(cols) #using sum() function to access the indexes of the columns that contain a number greater than 90
print("Indices of the column that contain a number greater than 90: ",indices)
```

#### (f) Which numbers appear in the matrix more than twice?

```python
flattened_array= arr.flatten() # using flatten() function to flatten a  multidimensional array into one dimension
unique_values, count = np.unique(flattened_array, return_counts = True) # using unique() function to find the unique values in the array and count the occurrence of each unique number in the array.
result = unique_values[count>2] #finding a number that appears more than twice
print("Numbers that appear more than twice in the matrix:" ,result)
```

#### (g) How many rows contain duplicate values?

```python
def check_duplicates(row): #defining a function to check whether any row contains duplicate values
    return len(row) != len(np.unique(row)) #using len() function to access and compare the length of the total number of elements in the row with the length of unique elements in the row
count_duplicates_in_row= np.sum([check_duplicates(row) for row in arr]) #check for duplicate values in the row by accessing each row in the array using for() loop and use sum() function to count duplicate values in a row
print("Number of rows that contain duplicate values: ",count_duplicates_in_row)
```
