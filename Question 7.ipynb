#### Question 7- Given the following matrix A and vector b:

```python
import numpy as np
```

##### a) Compute the determinant of matrix A

```python
A = np.array([
          [4,2,1],
          [3,1,5],
          [2,1,3]
])
determinant = np.linalg.det(A) #linalg is linear algebra module which includes det() function. This computes the determinant of the matrix
print("Determinant of matrix A:", determinant)
```

```python
b= np.array([10, 6, 7])
```

##### (b) Check if matrix A is invertible. If it is, find its inverse

```python
if determinant != 0:
    print("Matrix is invertible")
    inverse = np.linalg.inv(A) #use inv() function to compute the inverse of the matrix
    print("Inverse:\n" ,inverse)
else:
    print("Matrix is not invertible")
```

##### c) Solve the linear system Ax = b for x.

```python
x = np.linalg.solve(A, b) #use solve() function to compute the solution of the linear matrix equation Ax = b
print("Solution:" ,x)
```

##### (d) Compute the eigenvalues and eigenvectors of A.

```python
eigen_vals, eigen_vecs = np.linalg.eig(A) #use eig() function to compute the eigenvalues and eigenvectors of the matrix A
print("Eigen values:", eigen_vals)
print("Eigen vectors:\n",eigen_vecs)
```

##### e) Verify that the eigenvectors satisfy the equation Av = λv for each eigenvalue λ and eigenvector v.

```python
Av= A @ eigen_vecs #Av
lambdav= eigen_vals*eigen_vecs #λv
print("Av:\n", Av)
print("λv:\n", lambdav)
if np.allclose(Av, lambdav): # use allclose() function to compare two arrays element-wise and to check if they are "close" to each other.
    print("The equation Av = λv is satisfied for each eigenvalue λ and eigenvector v")
else:
    print("The equation Av = λv is not satisfied for each eigenvalue λ and eigenvector v")
```
