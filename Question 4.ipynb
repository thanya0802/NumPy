#### Question 4- Write a function that gets a 1-D array a and returns all the local maxima (peaks) in the array. 

```python
import numpy as np
```

```python
def local_maxima(a): #define local_maxima() function and pass variable a as a parameter. a is a list
    arr = np.array(a) #converting the list into NumPy array
    maxima = (arr[1:-1] >= arr[:-2]) & (arr[1:-1] >= arr[2:]) #arr[1:-1] selects all elements of the array except the first and last element and checks whether it is greater than or equal to their left neighbors(a[:-2]) and right neighbors(a[2:])
    return arr[1:-1][maxima]   #if the above condition is True, select and return the middle elements
```

```python
result= local_maxima([5, 8, 3, 1, 10, 9, 12]) #pass the list as arguments
print("Local maxima : ",result) 
```
