#### Question 3- Write a function that gets a 1-D array a and a number v, and returns the closest value in the array to v.

```python
import numpy as np
```

```python
a= np.array([10, 3, 7, 8, 15])
v= 4
def closest_val(a,v): #defining a function to return the closest value in the array to v
     difference= np.abs(a-v) #using abs() function to find the absolute difference between each element in the array and number v
     index= np.argmin(difference) #using argmin() function to find the index of minimum difference
     return a[index]
     
    
```

```python
result= closest_val(a,v) #call closest_val() function
print("Closest value: ", result)
```
